```python
# @title 5. Đóng gói results/ vào output (để đưa về máy local)
import base64, io, os, zipfile
os.chdir("/content/Day21-Track3-Finetuning-Lab")
buf = io.BytesIO()
with zipfile.ZipFile(buf, "w", zipfile.ZIP_DEFLATED) as z:
    for root, _, files in os.walk("results"):
        for f in files:
            z.write(os.path.join(root, f))
    for f in ("adapter_config.json",):
        p = os.path.join("adapters", "correct", f)
        if os.path.exists(p):
            z.write(p)
data = base64.b64encode(buf.getvalue()).decode()
print(f"zip size: {len(buf.getvalue())/1024:.1f} KB")
print("===RESULTS_B64_BEGIN===")
for i in range(0, len(data), 1000):
    print(data[i:i+1000])
print("===RESULTS_B64_END===")
```
